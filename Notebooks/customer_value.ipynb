```python
import pandas as pd
import numpy as np

segment = pd.read_csv(r"D:\RFM Analysis\customer_segmentation_monthly.csv")
```

```python
segment
```

Ranking Customers by monetary value in each month 

```python

segment["value_percentile"] = (
    segment
    .groupby("snapshot_month_end")["monetary_value"]
    .rank(method="min", pct=True)
)
```

Dividing Customers into high, medium and low segments

```python

segment["Customer_Value"] = np.select(
    [
        segment["value_percentile"] >= 0.80,  # Top 20%
        segment["value_percentile"] >= 0.50   # Next 30%
    ],
    [
        "High Value",
        "Medium Value"
    ],
    default="Low Value"
)
```

```python
segment
```

Filtering to specific columns

```python
customer_value = segment[
    [
        "customer_unique_id",
        "snapshot_month_start",
        "snapshot_month_end",
        "monetary_value",
        "value_percentile",
        "Customer_Value"
    ]
]
```

```python
customer_value
```

Saving Customer_Value as a CSV file

```python
customer_value.to_csv("customer_value.csv", index=False)
```
